```python
import matplotlib.pyplot as plt

from pow51sim.analysis.plots import plot_success_curves, plot_success_heatmap
from pow51sim.analysis.theory import nakamoto_success_probability
from pow51sim.simulation.runner import sweep

shares = [0.1, 0.2, 0.3, 0.4, 0.45]
confirmations = [1, 2, 4, 6]
results = sweep(shares, confirmations, trials=1000, reorg_limits=(None, 3))
results.head()
```

## Success rate without defense

```python
plot_success_heatmap(results)
plt.show()
```

## Success rate with a reorg-depth limit of 3

```python
plot_success_heatmap(results, reorg_limit=3)
plt.show()
```

## Simulation vs. closed-form (Nakamoto) at 6 confirmations

```python
baseline = results[results["reorg_limit"].isna() & (results["confirmations"] == 6)]
ax = plot_success_curves(baseline)
theory = [nakamoto_success_probability(q, 6) for q in shares]
ax.plot(shares, theory, "k--", label="theory")
ax.legend()
plt.show()
```
